```python
import numpy as np

docs = ['go INDIA',
		'INDIA INDIA',
		'Ise kehte hai hip hop',
		'Winner Winner Chiken Dinner',
		'bharat mata ki jai',
		'dhaakki Tikki',
		'BOB OGGY BOB',
		'OGGY the cat'
		'Andrew NG NG',
		'dhoni dhoni yes papa',
		'modi ki aisi taisi',
		'inquilab zindabad']
```

```python
from keras.preprocessing.text import Tokenizer
maketoken = Tokenizer(oov_token='<nothing>')
```

```python
maketoken.fit_on_texts(docs)
```

```python
maketoken.word_index
```

```python
maketoken.word_counts
```

```python
maketoken.document_count
```

```python
sequences = maketoken.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
```

```python
sequences
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train[0]
```

```python
len(X_train[2])
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(SimpleRNN(32,input_shape=(50,1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```
