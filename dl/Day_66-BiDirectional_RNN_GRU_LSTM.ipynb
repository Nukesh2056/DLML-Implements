```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import imdb
from tensorflow.keras.utils import pad_sequences   
from keras.models import Sequential
from keras.layers import Embedding, SimpleRNN, Dense, LSTM, GRU,Bidirectional
```

```python
# Define the BiDirectional RNN model
# Just use the bidirectional wrapper around the SimpleRNN layer to make it bidirectional, the params will get almost double after using bidirectional wrapper

model = Sequential([
    Embedding(10000, 32, input_length=100 ),
    Bidirectional(SimpleRNN(5, return_sequences=True)),
    Bidirectional(SimpleRNN(5)),
    Dense(1, activation='sigmoid')
    ])
model.summary() 
```

```python
# BiDirectional LSTM
model = Sequential([
    Embedding(10000, 32, input_length=100 ),
    Bidirectional(LSTM(5, return_sequences=True)),
    Bidirectional(LSTM(5)),
    Dense(1, activation='sigmoid')
    ])
model.summary() 
```

```python
# BiDirectional GRU
model = Sequential([
    Embedding(10000, 32, input_length=100 ),
    Bidirectional(GRU(5, return_sequences=True)),
    Bidirectional(GRU(5)),
    Dense(1, activation='sigmoid')
    ])
model.summary() 
```
