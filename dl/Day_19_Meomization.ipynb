```python
import time 
def feb(n):
    if n == 1 or n == 0:
        return 1 
    else : 
        return feb(n-1) + feb(n-2) 
```

```python
feb(32)
```

```python
import time 
def febM(n , d):
    if n in d :
        return d[n] 
    else : 
        d[n] = febM(n-1 , d ) + febM(n-2, d ) 
        return d[n] 
```

```python
d = { 0 : 1, 1 : 1 }

febM(100, d )
```
