```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten,AveragePooling2D
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
LeNET = Sequential()

LeNET.add(Conv2D(6, kernel_size=(5,5), padding='valid', activation='tanh', input_shape = (32,32,1)))
LeNET.add(AveragePooling2D(pool_size=(2,2), strides=2, padding='valid'))

LeNET.add(Conv2D(6, kernel_size=(5,5), padding='valid', activation='tanh'))
LeNET.add(AveragePooling2D(pool_size=(2,2), strides=2, padding='valid'))

LeNET.add(Flatten())

LeNET.add(Dense(120, activation='tanh'))
LeNET.add(Dense(84, activation='tanh'))
LeNET.add(Dense(10, activation='softmax'))

```
